```python
import json

exporters = []

cities = ["Karachi", "Lahore", "Faisalabad", "Sialkot", "Multan"]
products = ["denim fabric", "knitwear", "home textiles", "garments", "towels"]
certifications = ["OEKO-TEX", "GOTS", "GRS", "ISO 14001", "ZDHC", "Bluesign", "SA8000"]
markets = ["Germany", "France", "Italy", "Netherlands", "Sweden", "Spain"]

import random
random.seed(42)

for i in range(30):
    exporter = {
        "id": i + 1,
        "city": random.choice(cities),
        "product": random.choice(products),
        "export_market": random.choice(markets),
        "annual_revenue_usd": random.randint(500000, 5000000),
        "employees": random.randint(50, 2000),
        "current_certifications": random.sample(certifications, random.randint(0, 3)),
        "dpp_aware": random.choice([True, False])
    }
    exporters.append(exporter)

with open("dppbot/data/exporters.json", "w") as f:
    json.dump(exporters, f, indent=2)

print(f"✅ {len(exporters)} exporter profiles saved to dppbot/data/exporters.json")
print(f"Sample: {exporters[0]}")
```
